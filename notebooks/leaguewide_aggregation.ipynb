```python
# --- Setup ---
import pandas as pd
import numpy as np

# Load the full combo list (every possible combo per pitcher, not just
# the winners) that we exported from Phase 3
all_combos = pd.read_parquet("combos_handedness_only.parquet")

print(f"Loaded {len(all_combos)} combo rows")

# For each (pitcher, handedness), keep only their single BEST combo --
# same "lowest score wins" logic we've used throughout
best_vs_L = (
    all_combos.sort_values('combo_score_vs_L_weighted')
    .groupby('pitcher').first().reset_index()
    [['pitcher', 'pitch_1', 'pitch_2', 'pitch_3', 'combo_score_vs_L_weighted']]
    .rename(columns={'combo_score_vs_L_weighted': 'score'})
)
best_vs_L['stand'] = 'L'

best_vs_R = (
    all_combos.sort_values('combo_score_vs_R_weighted')
    .groupby('pitcher').first().reset_index()
    [['pitcher', 'pitch_1', 'pitch_2', 'pitch_3', 'combo_score_vs_R_weighted']]
    .rename(columns={'combo_score_vs_R_weighted': 'score'})
)
best_vs_R['stand'] = 'R'

# Combine into one table: one row per (pitcher, handedness) = 648 rows
best_overall = pd.concat([best_vs_L, best_vs_R], ignore_index=True)
print(f"Total (pitcher, handedness) best-combo rows: {len(best_overall)}")
best_overall.head()
```
